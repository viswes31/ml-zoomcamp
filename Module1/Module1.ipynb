Sum of Weights

```python
df1 = df[df['origin'] == 'Asia']
df1.head()
```

```python
array2d = df1[['vehicle_weight', 'model_year']]
array2d.shape
```

```python
array2d.head()
```

```python
X = array2d[0:7].to_numpy()
X
```

```python
XT= np.transpose(X)
```

```python
transposed = X.T
```

```python
res = X.T @ X
res
```

```python
inv = np.linalg.inv(res)
```

```python
inv
```

```python
inv.shape
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
w = np.matmul(np.matmul(inv, transposed), y)
```

```python
w
```
